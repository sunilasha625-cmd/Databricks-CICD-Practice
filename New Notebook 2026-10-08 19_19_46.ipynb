```python
%sql
SELECT 
    b1.room_id,
    b1.booking_id AS booking_1,
    b2.booking_id AS booking_2
FROM bookings b1
JOIN bookings b2
    ON b1.room_id = b2.room_id
    AND b1.booking_id < b2.booking_id   -- avoid duplicates
    AND b1.start_time < b2.end_time
    AND b1.end_time > b2.start_time;
```
